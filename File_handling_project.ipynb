```python
import csv
import os

if not os.path.exists("gradebook.csv"):
    with open("gradebook.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Name", "Python", "Software_engineering"])


while True:
    print("\n===== STUDENT GRADEBOOK =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter student name: ").strip()
        python_text = input("Enter Python marks: ").strip()
        Software_engineering_text = input("Enter Software_engineering marks: ").strip()
        # Do not accept empty marks
        if python_text == "" or Software_engineering_text == "":
            print("Marks cannot be empty.")
            continue

        try:
            python_marks = float(python_text)
            Software_engineering_marks = float(Software_engineering_text)

            # Check Python marks range
            if python_marks < 0 or python_marks > 100:
                print("Python marks must be between 0 and 100.")
                continue

            # Check Statistics marks range
            if Software_engineering_marks < 0 or Software_engineering_marks > 100:
                print("Software_engineering marks must be between 0 and 100.")
                continue

            # Save the student record in CSV
            with open("gradebook.csv", "a", newline="") as file:
                writer = csv.writer(file)
                writer.writerow([name, python_marks, Software_engineering_marks])

            print("Student record saved successfully.")

        except ValueError:
            print("Please enter numeric marks only.")

    elif choice == "2":
        try:
            with open("gradebook.csv", "r") as file:
                reader = csv.reader(file)

                print("\n===== STUDENT RECORDS =====")
                for row in reader:
                    print(row)

        except FileNotFoundError:
            print("Gradebook file does not exist.")

    elif choice == "3":
        print("Gradebook closed.")
        break

    else:
        print("Invalid choice. Please select 1, 2, or 3.")
```
